```python
import torch
```

```python
B = 1
T = 2
n_heads = 3
d_head = 5
data = torch.arange(B*T*n_heads*d_head).reshape(B, T, n_heads * d_head)
```

```python
data.stride(), data
```

```python
d1 = data.view(B, n_heads, T, d_head)
d2 = data.view(B, T, n_heads, d_head).permute(0, 2, 1, 3)
print(d1.shape, d2.shape, d1.shape == d2.shape)
print(d1, d1.stride())
print(d2, d2.stride())
```
